# A - Задача на вычисление

В общежитии института в каждой комнате можно поселить четырех человек. Какое наименьшее количество комнат необходимо для поселения 83 иногородних студентов?

## Задание

Напишите программу на языке Python для решения задачи с использованием переменных.

```python
# Ваш код здесь
# 1. Входные данные. Здесь нужно определить переменные для хранения входных данных и присвоить им значения.
students = 83
places = 4

# 2. Обработка данных. Здесь нужно вычислить искомое значение и сохранить его в переменной.
rooms = students // places

if students % places != 0:
    rooms += 1


# 3. Вывод результатов. Здесь нужно вывести вычисленное значение на экран.
print("Необходимое количество комнат:", rooms)
```

# B - Задача на ввод и вывод

На счету Машиного мобильного телефона было X рублей, а после разговора с Леной осталось Y рублей. Сколько минут длился разговор с Леной, если одна минута разговора стоит 2 рубля 50 копеек?

## Задание

Напишите программу на языке Python для решения задачи с использованием переменных.

```python
# Ваш код здесь
# 1. Входные данные. Здесь нужно определить переменные для хранения входных данных и ввести значения с клавиатуры.
X = float(input("Сколько рублей было на счету? "))
Y = float(input("Сколько рублей осталось после разговора? "))

# 2. Обработка данных. Здесь нужно вычислить искомое значение и сохранить его в переменной.
cost = 2.5
minutes = (X - Y) / cost

# 3. Вывод результатов. Здесь нужно вывести вычисленное значение на экран.
print("Разговор длился", minutes, "минут")
```

# C - Задача про Рудольфа

Рудольф меняет ремень по настроению. При мрачном настроении он надевает ремень из крокодильей кожи, при хорошем - из бычьей, при плохом - из свиной, при веселом - из кожезаменителя

## Задание

Напишите программу на языке Python, которая по введенному настроению выведет для Рудольфа тип ремня (тип кожи).

```python
# Ваш код здесь
# 1. Входные данные. Здесь нужно определить переменные для хранения входных данных и ввести значения с клавиатуры.
mood = input("Настроение Рудольфа: ")
# 2. Обработка данных. Здесь нужно вычислить искомое значение и сохранить его в переменной.
if mood == "мрачное":
    belt = "ремень из крокодильей кожи"

elif mood == "хорошее":
    belt = "ремень из бычьей кожи"

elif mood == "плохое":
    belt = "ремень из свиной кожи"

elif mood == "веселое":
    belt = "ремень из кожезаменителя"
# 3. Вывод результатов. Здесь нужно вывести вычисленное значение на экран.
print("Рудольф наденет:", belt)
```

#### Задание 7: Найдите среднее арифметическое нечетных элементов списка

```python
# Ваш код здесь
lst = [15, -48, -30, 17, 8, -29, -22, 39, 20, 1, -35, 0, 44, 26, -9, 15, 11, -25, -6, -32]

sum = 0 #сумма
count = 0 #количество

for num in lst:
    if num % 2 != 0:
        sum += num
        count += 1

average = sum / count

print("Среднее арифметическое нечетных элементов:", average)
```
