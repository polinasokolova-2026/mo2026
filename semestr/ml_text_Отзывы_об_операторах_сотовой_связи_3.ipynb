#### 11. Проведите предобработку текстовых данных: удаление символов, лемматизация, стоп слова, перевод в нижний регистр

```python
!pip install pymystem3
```

```python
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

from pymystem3 import Mystem
```

```python
# Ваш код здесь
# Загрузка стоп-слов для русского языка
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

# Инициализация Mystem
mystem = Mystem()

# Функция для предобработки текста
def preprocess_text(text):
    # Приведение к нижнему регистру
    text = text.lower()

    # Удаление пунктуации и цифр
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\d+', '', text)

    # Токенизация текста
    tokens = word_tokenize(text, language='russian')

    # Удаление стоп-слов
    stop_words = set(stopwords.words('russian'))
    tokens = [word for word in tokens if word not in stop_words]

    # Лемматизация с использованием Mystem
    lemmas = mystem.lemmatize(' '.join(tokens))
    lemmas = [lemma.strip() for lemma in lemmas if lemma.strip()]

    return ' '.join(lemmas)

data['processed_review'] = data['phrase'].apply(preprocess_text)

data.head()
```

# Часть II. Построение модели машинного обучения для набора данных: Отзывы об операторах сотовой связи

## Задание

#### 13. Разделить выборку на признаки (Х) и результат (Y)

```python
# Ваш код здесь
from sklearn.model_selection import train_test_split
X = data['processed_review']
Y = data['telecom']
```

#### 14. Разделить на обучающую и тестовую выборки

```python
# Ваш код здесь
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.33, random_state=42)

print(x_train.shape)
print(x_test.shape)
```

#### 15. TF-IDF векторизация и мешок слов

```python
# Ваш код здесь

from sklearn.neighbors import KNeighborsClassifier
#Векторизация текста
from sklearn.feature_extraction.text import CountVectorizer

from nltk import ngrams
```

```python
vec = CountVectorizer(ngram_range=(1, 1)) #строим BoW для слов
bow = vec.fit_transform(x_train) #преобразуем обучающие данные
```

#### 16. Сформировать модель машинного обучения

```python
# Ваш код здесь

clf = KNeighborsClassifier(n_neighbors=5)
```

#### 17. Обучить модель

```python
# Ваш код здесь
clf.fit(bow, y_train)
```

#### 18. Оценить качество модели

```python
from sklearn.metrics import accuracy_score, classification_report

# преобразуем тестовые данные
x_test_bow = vec.transform(x_test)
y_pred = clf.predict(x_test_bow)

# точность модели
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

#### 19. Выполнить предсказание класса для трех разных фраз

```python
phrases = [
    'У меня плохо работает интернет от МТС',
    'Билайн опять списал деньги',
    'Ростелеком радует качеством связи'
]

phrases_processed = [preprocess_text(text) for text in phrases]

phrases_vectorized = vec.transform(phrases_processed)

predictions = clf.predict(phrases_vectorized)

for phrase, pred in zip(phrases, predictions):
    print(phrase)
    print(pred)
```

#### 21. Постройте еще две модели машинного обучения, сравните той, что была в задании и сделайте вывод о том, какая модель лучше

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
```

```python
#Logistic Regression

lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(bow, y_train)
lr_pred = lr_model.predict(x_test_bow)
print(accuracy_score(y_test, lr_pred))

print(classification_report(y_test, lr_pred))
```

```python
# Проверка работы Logistic Regression на конкретных примерах

from sklearn.feature_extraction.text import CountVectorizer

count_vec = CountVectorizer(ngram_range=(1,1))

bow = count_vec.fit_transform(x_train)

x_test_bow = count_vec.transform(x_test)

examples = ["МТС постоянно присылает рекламу и навязывает услуги",
    "Билайн радует быстрым интернетом",
    "Мегафон плохо ловит связь в метро"]

examples_processed = [preprocess_text(text) for text in examples]

examples_vectorized = count_vec.transform(examples_processed)

predictions = lr_model.predict(examples_vectorized)


for text, pred in zip(examples, predictions):
    print(text)
    print(pred)
```

```python
#Naive Bayes

nb_model = MultinomialNB()
nb_model.fit(bow, y_train)
nb_pred = nb_model.predict(x_test_bow)
nb_acc = accuracy_score(y_test, nb_pred)

print(nb_acc)
print(classification_report(y_test, nb_pred))
```

```python
# Проверка работы модели Naive Bayes на конкретных примерах

examples = ["МТС постоянно присылает рекламу и навязывает услуги",
    "Билайн радует быстрым интернетом",
    "Мегафон плохо ловит связь в метро"]

examples_processed = [preprocess_text(text) for text in examples]

examples_vectorized = count_vec.transform(examples_processed)
nb_predictions = nb_model.predict(examples_vectorized)

for text, pred in zip(examples, nb_predictions):
    print(text)
    print(pred)
```
